```python
import pandas as pd
from sqlalchemy import create_engine

file_path = "/home/aneesh/Downloads/employee_dataset/employee_records.csv"
chunk_size = 100
connection_string = "postgresql://rag:rag@localhost:5432"

# read_csv returns a data generator
data = pd.read_csv(file_path, chunksize=chunk_size)
# next(data).head()

# create a connection
engine = create_engine(connection_string)

for chunk in data:
    chunk.to_sql(
        "employee_records", 
        engine, 
        if_exists="append", 
        index=False
    )
```
