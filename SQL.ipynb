## Create
- database
```sql
CREATE DATABSE name;
USE name;
```

- table
    - syntax
```SQL
CREATE TABLE table_name (
    column1 datatype constraints,
    column2 datatype constraints,
    column3 datatype constraints
);
```

    - example
```SQL
CREATE TABLE employees (
    employee_id INT PRIMARY KEY,
    first_name VARCHAR(50),
    last_name VARCHAR(50),
    hire_date DATE
);
```

- CONSTRAINS
    - NOT NULL
    - UNIQUE   
        - unique values
        - one NULL accepted
    - CHECK
        - safety consideration, if not supplied the value agreeing the condition, the row will be rejected
        - INT CHECK (age >= 18)
    - DEFAULT
        - provide a default value
        - `status VARCHAR(20) DEFAULT 'ACTIVE'`
    - PRIMARY KEY
        - combination of NOT NULL and UNIQUE
        - `id INT PRIMARY KEY`
    - FOREIGN KEY
        - a primary key in another table
        - enforcing referential integrity
        - `FOREIGN KEY (department_id) REFERENCES departments(id)`

    ```SQL
    CREATE TABLE employees (
    employee_id INT PRIMARY KEY,                       -- Primary Key
    first_name VARCHAR(50) NOT NULL,                   -- Not Null
    email VARCHAR(100) UNIQUE,                         -- Unique
    age INT CHECK (age >= 18),                         -- Check
    hire_date DATE DEFAULT CURRENT_DATE,               -- Default
    department_id INT,
    CONSTRAINT fk_dept FOREIGN KEY (department_id)     -- Foreign Key
        REFERENCES departments(id)
    );
    ```

    - example
    ```sql
    create table temp(
    id INT Primary key,
    name varchar(50) not null,
    email varchar(50) unique,
    age int check (age >= 0 AND age <= 120)
    );
    ```

## cte

``` sql
WITH min_s AS (
    SELECT
        "Country",
        "Position",
        MIN("Salary") AS min_salary
    FROM employee_records
    GROUP BY "Country", "Position"
),
max_s AS (
    SELECT
        "Country",
        "Position",
        MAX("Salary") AS max_salary
    FROM employee_records
    GROUP BY "Country", "Position"
)
SELECT
    min_s."Country",
    min_s."Position",
    min_s.min_salary,
    max_s.max_salary,
    (max_s.max_salary - min_s.min_salary) AS diff
FROM min_s
JOIN max_s
    ON min_s."Country" = max_s."Country"
    AND min_s."Position" = max_s."Position"
ORDER BY diff DESC;

```

## Window functions
-
